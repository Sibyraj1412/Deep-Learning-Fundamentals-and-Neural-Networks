# Notebook 6: Loss Functions

In Forward Propagation, the model makes a prediction. But how do we know if that prediction is good or bad? We use a **Loss Function** (also called a Cost Function) to measure the "distance" between the model's prediction and the actual target.

**Learning Methodology:**
**Definition $\rightarrow$ Formula $\rightarrow$ Intuition $\rightarrow$ Implementation $\rightarrow$ Use Case**

## Why are Loss Functions Required?

**Explanation:**
A loss function provides a single numerical value that represents the error of the model. This value is the "signal" the model uses to learn.
- **High Loss:** The model is very wrong.
- **Low Loss:** The model is very accurate.

The goal of training a neural network is to **minimize the loss function** by adjusting weights and biases.

## 1. Regression Loss Functions
Used when the target is a continuous number (e.g., predicting house prices).

### A. Mean Absolute Error (MAE)
**Definition:** The average of the absolute differences between predictions and targets.
**Formula:** $\text{MAE} = \frac{1}{n} \sum |y_{true} - y_{pred}|$
**Intuition:** "On average, how many units is the model off by?"
**Pros:** Robust to outliers.
**Cons:** Gradient is constant, making it harder to converge precisely.

```python
import numpy as np
import torch
import torch.nn as nn

y_true = np.array([100, 200, 300])
y_pred = np.array([110, 190, 250])

# NumPy implementation
mae_np = np.mean(np.abs(y_true - y_pred))
print(f"MAE (NumPy): {mae_np:.4f}")

# PyTorch implementation
y_true_t = torch.tensor(y_true, dtype=torch.float32)
y_pred_t = torch.tensor(y_pred, dtype=torch.float32)
mae_torch = torch.mean(torch.abs(y_true_t - y_pred_t))
print(f"MAE (PyTorch): {mae_torch.item():.4f}")
```

### B. Mean Squared Error (MSE)
**Definition:** The average of the squared differences.
**Formula:** $\text{MSE} = \frac{1}{n} \sum (y_{true} - y_{pred})^2$
**Intuition:** Squaring the error penalizes **large errors** much more than small ones.
**Pros:** Mathematically smooth (easy to calculate gradients).
**Cons:** Extremely sensitive to outliers.

```python
# NumPy implementation
mse_np = np.mean((y_true - y_pred)**2)
print(f"MSE (NumPy): {mse_np:.4f}")

# PyTorch implementation
mse_torch = nn.MSELoss()(y_pred_t, y_true_t)
print(f"MSE (PyTorch): {mse_torch.item():.4f}")
```

### C. Root Mean Squared Error (RMSE)
**Definition:** The square root of MSE.
**Formula:** $\text{RMSE} = \sqrt{\text{MSE}}$
**Intuition:** Brings the error units back to the original scale of the data.

```python
rmse_np = np.sqrt(mse_np)
print(f"RMSE (NumPy): {rmse_np:.4f}")
```

## 2. Binary Classification Loss
Used when the target is binary (0 or 1).

### Binary Cross Entropy (BCE)
**Definition:** Measures the performance of a classification model whose output is a probability value between 0 and 1.
**Formula:** $\text{BCE} = -\frac{1}{n} \sum [y \log(\hat{y}) + (1-y) \log(1-\hat{y})]$
**Intuition:** It heavily penalizes predictions that are "confident and wrong" (e.g., predicting 0.99 for a target of 0).

**BCE with Logits:** In PyTorch, `BCEWithLogitsLoss` combines a Sigmoid layer and BCE loss into one class for better numerical stability.

```python
# Target: [1, 0], Predictions: [0.9, 0.1] (Very accurate)
y_true_bin = torch.tensor([1.0, 0.0])
y_pred_bin = torch.tensor([0.9, 0.1])

bce_loss = nn.BCELoss()(y_pred_bin, y_true_bin)
print(f"BCE Loss (Accurate): {bce_loss.item():.4f}")

# Target: [1, 0], Predictions: [0.1, 0.9] (Very wrong)
y_pred_wrong = torch.tensor([0.1, 0.9])
bce_loss_wrong = nn.BCELoss()(y_pred_wrong, y_true_bin)
print(f"BCE Loss (Wrong): {bce_loss_wrong.item():.4f}")
```

## 3. Multiclass Classification Loss

### Cross Entropy Loss
**Definition:** Generalization of BCE for multiple classes.
**Formula:** $\text{CE} = -\sum y_{true} \log(\hat{y}_{pred})$
**Intuition:** It compares the predicted probability distribution with the true one-hot encoded label.

**Implementation Note:** PyTorch's `nn.CrossEntropyLoss` expects **raw logits** (unnormalized scores) and applies Softmax internally.

```python
# 3 classes: Cat, Dog, Bird
# True label: Dog (Index 1)
y_true_multi = torch.tensor([1]) 
# Raw scores from model (logits)
y_pred_multi = torch.tensor([[1.2, 4.5, -1.1]]) 

ce_loss = nn.CrossEntropyLoss()(y_pred_multi, y_true_multi)
print(f"Cross Entropy Loss: {ce_loss.item():.4f}")
```

## Summary: Prediction $\rightarrow$ Loss $\rightarrow$ Learning

1.  **Prediction:** The model takes input $\rightarrow$ Forward Propagation $\rightarrow$ Prediction $\hat{y}$.
2.  **Loss:** The loss function compares $\hat{y}$ to the true $y$ $\rightarrow$ Error Value.
3.  **Learning:** The model uses the error value to calculate gradients (Backpropagation) and updates weights (Optimizer) to make the loss smaller next time.
