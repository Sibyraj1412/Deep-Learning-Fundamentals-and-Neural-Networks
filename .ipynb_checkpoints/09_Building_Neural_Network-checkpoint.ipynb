# Notebook 9: Building a Neural Network with PyTorch

In the previous notebooks, we learned the a single neuron and manual propagation. Now, we will learn how to use PyTorch's high-level API to build a full-scale Artificial Neural Network (ANN).

**Learning Methodology:**
**Concept $\rightarrow$ Implementation $\rightarrow$ Inspection $\rightarrow$ Interpretation**

## 1. Understanding `nn.Module`

**Explanation:**
In PyTorch, every neural network must inherit from `nn.Module`. Think of `nn.Module` as a blueprint. It provides the necessary infrastructure to track parameters, move the model to a GPU, and handle the forward pass.

**The Two Main Components:**
1.  **The Constructor (`__init__`):** This is where you define the "parts" of your network (the layers). You decide how many layers you want and how many neurons are in each.
2.  **The Forward Method (`forward`):** This is where you define the "flow". You specify how the input data travels through the layers you defined in the constructor.

## 2. Building the Network Architecture

We will build a network with the following flow:
**Input $\rightarrow$ Hidden Layer 1 $\rightarrow$ Activation $\rightarrow$ Hidden Layer 2 $\rightarrow$ Output**

**Key Terms:**
- `nn.Linear(in_features, out_features)`: A fully connected layer that performs $y = xW^T + b$.
- **Input Features:** The number of columns in your data.
- **Hidden Neurons:** The number of "filters" in each layer. More neurons can capture more complex patterns but may overfit.
- **Output Neurons:** Usually 1 for binary classification or $N$ for $N$-class classification.

```python
import torch
import torch.nn as nn

class MyFirstNeuralNet(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(MyFirstNeuralNet, self).__init__()
        
        # Layer 1: Input to Hidden 1
        self.layer1 = nn.Linear(input_size, hidden_size)
        self.relu1 = nn.ReLU()
        
        # Layer 2: Hidden 1 to Hidden 2
        self.layer2 = nn.Linear(hidden_size, hidden_size)
        self.relu2 = nn.ReLU()
        
        # Layer 3: Hidden 2 to Output
        self.layer3 = nn.Linear(hidden_size, output_size)
        self.sigmoid = nn.Sigmoid() # For binary classification

    def forward(self, x):
        # The data flow
        x = self.layer1(x)
        x = self.relu1(x)
        x = self.layer2(x)
        x = self.relu2(x)
        x = self.layer3(x)
        x = self.sigmoid(x)
        return x

# Initialize the model
# Scenario: 10 input features, 16 hidden neurons, 1 output
input_dim = 10
hidden_dim = 16
output_dim = 1
model = MyFirstNeuralNet(input_dim, hidden_dim, output_dim)

print(model)
```

## 3. Inspecting Model Parameters

**What are parameters?**
Parameters are the **Weights** and **Biases** that the model learns during training.

- **Weight Matrix:** Stores the connection strength between neurons of two layers.
- **Bias Vector:** Stores the offset for each neuron.
- **Trainable Parameters:** Parameters that are updated via backpropagation.

```python
# 1. Print total parameters
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

print(f"Total Parameters: {total_params}")
print(f"Trainable Parameters: {trainable_params}")

# 2. Inspect weights and biases of the first layer
print("\nLayer 1 Weights Shape:", model.layer1.weight.shape) # (out_features, in_features)
print("Layer 1 Bias Shape:", model.layer1.bias.shape)       # (out_features)

# 3. View a slice of the weights
print("\nFirst 5 weights of the first neuron in Layer 1:\n", model.layer1.weight[0][:5])
```

## 4. Summary and Interpretation

**How to read the architecture:**
- If `layer1` is `nn.Linear(10, 16)`, it means the model takes 10 numbers and turns them into 16. This requires $10 \times 16 = 160$ weights and $16$ biases.
- The **Activation layers** (ReLU) don't have parameters; they just apply a mathematical rule to the output of the linear layers.
- The final **Sigmoid** ensures the output is a probability between 0 and 1.
