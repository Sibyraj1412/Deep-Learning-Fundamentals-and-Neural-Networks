## 1. Biological Neuron vs Artificial Neuron

**Explanation:**
Deep Learning is inspired by the human brain.
- **Biological Neuron:** Receives electrical signals through *dendrites*, processes them in the *cell body*, and if the signal is strong enough, fires an electrical pulse through the *axon*.
- **Artificial Neuron:** Receives *numerical inputs*, processes them using *weights and biases*, and if the result is above a certain threshold, produces a *numerical output*.

**Comparison Table:**
| Biological Component | Artificial Equivalent |
| :--- | :--- |
| Dendrites | Inputs ($x$) |
| Synaptic Strength | Weights ($w$) |
| Cell Body Threshold | Bias ($b$) |
| Axon | Output ($y$) |

## 2. Anatomy of a Perceptron

**Concept:**
A perceptron performs two main steps:
1. **The Weighted Sum:** It multiplies every input by a weight and adds a bias.
2. **The Activation:** It passes that sum through a function (usually a Step Function) to decide the output.

**Mathematical Representation:**
First, calculate the linear combination:
$$z = (w_1 \cdot x_1) + (w_2 \cdot x_2) + \dots + (w_n \cdot x_n) + b$$

Then, apply the activation:
$$\text{output} = \text{activation}(z)$$

**Analogy:**
Imagine deciding whether to go to the beach.
- Inputs: `[Is it sunny?, Is it the weekend?]`
- Weights: `[Sunny=10, Weekend=5]` (Sunny is more important).
- Bias: `-7` (Your general reluctance to leave the house).
- If the total sum $> 0$, you go to the beach!

```python
import numpy as np

def perceptron(inputs, weights, bias):
    # Step 1: Weighted Sum (z = w*x + b)
    z = np.dot(inputs, weights) + bias
    # Step 2: Activation (Simple Step Function)
    return 1 if z > 0 else 0

# Test inputs: [Sunny, Weekend]
x = np.array([1, 1]) # Both True
w = np.array([10, 5]) # Weights
b = -7               # Bias

print(f"Decision: {perceptron(x, w, b)}") # Output: 1 (Yes, go to beach)
```

## 3. Implementing Logic Gates

A single perceptron can represent basic logical operations if we set the weights and biases correctly.

### The AND Gate
The AND gate only outputs 1 if **both** inputs are 1.
- Logic: If $x_1=1$ AND $x_2=1$, then $y=1$.

```python
def and_gate(x1, x2):
    inputs = np.array([x1, x2])
    weights = np.array([1, 1])
    bias = -1.5 # Only 1+1 (2) is greater than 1.5
    return perceptron(inputs, weights, bias)

print("AND Gate Results:")
for x1, x2 in [(0,0), (0,1), (1,0), (1,1)]:
    print(f"{x1} AND {x2} = {and_gate(x1, x2)}")
```

### The OR Gate
The OR gate outputs 1 if **at least one** input is 1.
- Logic: If $x_1=1$ OR $x_2=1$, then $y=1$.

```python
def or_gate(x1, x2):
    inputs = np.array([x1, x2])
    weights = np.array([1, 1])
    bias = -0.5 # Any 1 is greater than 0.5
    return perceptron(inputs, weights, bias)

print("OR Gate Results:")
for x1, x2 in [(0,0), (0,1), (1,0), (1,1)]:
    print(f"{x1} OR {x2} = {or_gate(x1, x2)}")
```

## 4. The Limitation of a Single Perceptron: The XOR Problem

**The Problem:**
The XOR (Exclusive OR) gate outputs 1 if **exactly one** input is 1, but 0 if both are 0 or both are 1.

**Why it fails:**
A single perceptron is a **linear classifier**. It can only draw one straight line to separate two classes. If you plot the points of an XOR gate, you will see that it is impossible to separate the 0s from the 1s with a single straight line.

**The Solution:**
To solve XOR, we need **Multiple Layers**. By stacking neurons in layers (creating a Multi-Layer Perceptron), we can create "curved" or "complex" decision boundaries.

**Interpretation:**
This is exactly why "Deep" Learning exists. One neuron is too simple; a network of neurons is powerful.
