# Notebook 7: Backpropagation & Gradients

Forward propagation tells us how wrong the model is (the Loss). **Backpropagation** is the process of figuring out *which* weights and biases were responsible for that error and by *how much*.

**Learning Methodology:**
**Concept $\rightarrow$ Math $\rightarrow$ PyTorch Implementation $\rightarrow$ Interpretation**

## 1. What is a Gradient?

**Explanation:**
A gradient is simply a partial derivative. In Deep Learning, the gradient of the loss function with respect to a weight ($\frac{\partial \text{Loss}}{\partial w}$) tells us:
*"If I increase this weight by a tiny amount, how much will the loss change?"*

- **Positive Gradient:** Increasing the weight increases the loss $\rightarrow$ **Decrease the weight**.
- **Negative Gradient:** Increasing the weight decreases the loss $\rightarrow$ **Increase the weight**.
- **Zero Gradient:** We are at a minimum (or a plateau).

## 2. Gradient Descent & The Computational Graph

**Concept:**
Imagine you are on a foggy mountain and want to reach the valley (the lowest loss). You can't see the whole mountain, but you can feel the slope under your feet. You take a step in the direction of the steepest descent.

**The Computational Graph:**
Deep Learning frameworks represent calculations as a graph.
`Input` $\rightarrow$ `Op1` $\rightarrow$ `Op2` $\rightarrow$ `Loss`.
When we go backward, we simply traverse this graph in reverse to distribute the "blame" for the error.

## 3. The Chain Rule: The Heart of Backpropagation

**Explanation:**
If $y$ depends on $u$, and $u$ depends on $x$, then the rate of change of $y$ with respect to $x$ is the product of the rates of change:
$$\frac{dy}{dx} = \frac{dy}{du} \cdot \frac{du}{dx}$$

In a neural network, the loss depends on the output, the output depends on the hidden layer, and the hidden layer depends on the weights. The chain rule allows us to propagate the error from the end back to the beginning.

## 4. Manual Gradient Example

Let's use a tiny function: $y = w \cdot x$ (no bias, no activation).
- $x = 2.0$
- $w = 3.0$
- Prediction $\hat{y} = 3 \cdot 2 = 6.0$
- Target $y = 10.0$
- Loss $L = (\hat{y} - y)^2 = (6 - 10)^2 = 16.0$

**Calculate Gradient $\frac{\partial L}{\partial w}$:**
1. $\frac{\partial L}{\partial \hat{y}} = 2(\hat{y} - y) = 2(6 - 10) = -8$
2. $\frac{\partial \hat{y}}{\partial w} = x = 2$
3. By Chain Rule: $\frac{\partial L}{\partial w} = -8 \cdot 2 = -16$

**Interpretation:** Since the gradient is $-16$ (negative), we should **increase** $w$ to decrease the loss.

```python
import torch

# 1. Input and Target
x = torch.tensor([2.0])
target = torch.tensor([10.0])

# 2. Weight with gradient tracking enabled
w = torch.tensor([3.0], requires_grad=True)

# 3. Forward Pass
y_pred = w * x
loss = (y_pred - target)**2

print(f"Prediction: {y_pred.item():.2f}, Loss: {loss.item():.2f}")

# 4. Backward Pass
loss.backward()

# 5. Access the gradient
print(f"Gradient of loss w.r.t weight w: {w.grad.item():.2f}")
```

## 5. How PyTorch's Autograd Works

PyTorch uses a system called **Autograd** to automate the chain rule.
- **`requires_grad=True`**: Tells PyTorch to build a computational graph for this tensor.
- **`.backward()`**: Starts the backpropagation process from the loss tensor. It calculates the gradient for every tensor in the graph that has `requires_grad=True`.
- **`.grad`**: Stores the resulting gradient.

**The Loop:**
`Forward Pass` $\rightarrow$ `Calculate Loss` $\rightarrow$ `Backward Pass` $\rightarrow$ `Calculate Gradients` $\rightarrow$ `Update Weights`
